```python
import pandas as pd 
```

```python
claims = pd.read_json (r"D:\DEPI\DEPI final project\data sources\claims.json")
```

```python
payers = claims["payer_name"].drop_duplicates().reset_index(drop=True)
payers_df = pd.DataFrame({
    "payer_id": range(1, len(payers)+1),
    "payer_name": payers
})
```

```python
payers_df.head()
```

```python
claims = claims.merge(payers_df, on="payer_name")
```

```python
claims = claims.drop(columns=["payer_name"])
```

```python
claims.to_json(r"D:\DEPI\DEPI final project\data sources\claims_Table.json", orient="records", lines=True)
```
